```python
def LU(A):
    
    # Find dimension of A
    n = A.shape[0]
    
    # Initialise L=I, U=A
    L = np.eye(n) 
    U = np.copy(A)

    for k in range(n - 1): # loop over columns 1 to n-1
        for j in range(k + 1, n): # loop over rows k+1 to n
            L[j, k] = U[j, k] / U[k ,k] # compute the multiplier l_jk
            U[j, k:] = U[j, k:] - L[j, k] * U[k, k:] 
            # subtract a multiple of row k 
            # from row j to create zeros 
            # below the diagonal in column k
    
    return L, U # return the LU factorisation of A
```

A numerical method for the solution of partial differential equations is based on the least-square algorithm. In this question, you will use it to compute the fluid flow around an obstacle in two dimensions using two simple models.

**The first model is for inviscid fluids.** The velocity field outside the obstacle, $\mathbf{u}(\mathbf{x})$ with $\mathbf{x}=(x,y)$, is given by
$$
\mathbf{u} = - U \mathbf{i} + \nabla \phi,
$$
where $U$ is a constant and $\mathbf{i}$ is the unit vector in the $x$-direction. The function $\phi(x,y)$ must satisfy Laplace's equation
$$
\nabla^2 \phi = \partial_{xx} \phi + \partial_{yy} \phi = 0.
$$
The term $\nabla \phi$ decays as $\|\mathbf{x}\|_2 \to \infty$ so the velocity tends to the uniform incoming velocity $-U \mathbf{i}$. We set $U=1$ for simplicity. 

The function $\phi(\mathbf{x})$ is determined by a condition imposed on the boundary of the obstacle. For an inviscid fluid, it is simply that the flow does not penetrate the obstacle.
This requires that $\mathbf{n} \cdot \mathbf{u} = 0$, that is,
$$
\mathbf{n} \cdot \nabla \phi = n_1 \tag{1}
$$
on the boundary of the obstacle. Here $\mathbf{n}=(n_1,n_2)$ is the normal to the boundary. 

(a) Check that $\phi(x,y) = \ln \| \mathbf{x} - \hat{\mathbf{x}} \|_2$ for fixed $\hat{\mathbf{x}}$ satisfies Laplace's equation. Deduce that 
$$
\phi(\mathbf{x}) = \sum_{j=1}^n c_j \ln \| \mathbf{x} - \hat{\mathbf{x}}_j \|_2 \tag{2}
$$
satisfies Laplace's equation. 
Here $\hat{\mathbf{x}}_j, \, j=1,\cdots,n,$ are a collection of points inside the obstacle and the $c_j$ are constants.

**[6 marks]**

We find a numerical approximation to $\phi$ by assuming it takes the form (2). The constant $c_j$ are determined by requiring that the boundary condition (1) be approximately satisfied on points $\mathbf{x}_i, \, i=1,\cdots,m$ on the boundary. 
Specifically, we solve the minimisation problem
$$
\min_{c_1,\cdots,c_n} \sum_{i=1}^m \left( (\mathbf{n} \cdot \nabla \phi - n_1 )|_{\mathbf{x}=\mathbf{x}_i} \right)^2. \tag{3}
$$
We take $m > n$. 

(b) Recast this minimisation problem in the standard least-square form
$$
